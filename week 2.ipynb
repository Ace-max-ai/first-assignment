# Interpreted vs Compiled Languages

An interpreted language executes code using an interpreter.
Python is a common example.

A compiled language translates code into machine code
or another target form before execution.
C and C++ are common examples.

The main difference is how the code is translated
and executed.

```python
age = 20
price = 9.99
name = "Alex"
is_student = True
number = 2 + 3j

print(type(age))
print(type(price))
print(type(name))
print(type(is_student))
print(type(number))
```
